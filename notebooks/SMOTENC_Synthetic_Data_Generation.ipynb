```python
dataset['Systolic Pressure'] = dataset['Blood Pressure'].apply(lambda x: int(x.split('/')[0]))
dataset['Diastolic Pressure'] = dataset['Blood Pressure'].apply(lambda x: int(x.split('/')[1]))
```

```python
dataset.head()
```

```python
dataset.drop('Blood Pressure',inplace=True,axis=1)
dataset.drop('Person ID',inplace=True,axis=1)
```

```python
dataset.info()
```

```python
categorical_features = ['Gender','Occupation','BMI Category', 'Sleep Disorder']
```

```python
for cat in categorical_features:
  base = 0
  c_map = {}

  def map(x):
    global base

    if x in c_map.keys():
      return c_map[x]
    else:
      c_map[x] = base
      base+=1
      return c_map[x]

  dataset[cat] = dataset[cat].apply(map)

  print(cat," -> ",c_map)
```

```python
X = dataset.drop('Sleep Disorder', axis=1)
y = dataset['Sleep Disorder']

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print(y.value_counts())
```

0 → Healthy → 375

1 → Sleep Apnea → 93

2 → Insomnia → 91

```python
target_total = 1500

healthy = round(375 / 559 * target_total)
sleep_apnea = round(91 / 559 * target_total)
insomnia = target_total - healthy - sleep_apnea

print("Healthy:", healthy)
print("Sleep Apnea:", sleep_apnea)
print("Insomnia:", insomnia)
print("Total:", healthy + sleep_apnea + insomnia)
```

```python
sampling_strategy = {
    0: 1006,
    1: 244,
    2: 250
}
```

```python
categorical_features = [0, 2, 7]
```

```python
from imblearn.over_sampling import SMOTENC

smotenc = SMOTENC(
    categorical_features=categorical_features,
    random_state=42,
    sampling_strategy=sampling_strategy
)
```

```python
X_resampled, y_resampled = smotenc.fit_resample(X, y)
```

```python
print("X shape:", X_resampled.shape)
print("y shape:", y_resampled.shape)

print(y_resampled.value_counts())
```

```python
new_dataset = pd.DataFrame(X_resampled, columns=X.columns)

new_dataset['Sleep Disorder'] = y_resampled

print(new_dataset.shape)
new_dataset.head()
```

```python
new_dataset.insert(0, 'Person ID', range(1, len(new_dataset) + 1))
```

BMI Category  ->  {'Overweight': 0, 'Normal': 1, 'Obese': 2, 'Normal Weight': 3}

Sleep Disorder  ->  {'Healthy': 0, 'Sleep Apnea': 1, 'Insomnia': 2}

```python
gender_map = {
    0: 'Male',
    1: 'Female'
}

occupation_map = {
    0: 'Software Engineer',
    1: 'Doctor',
    2: 'Sales Representative',
    3: 'Teacher',
    4: 'Nurse',
    5: 'Engineer',
    6: 'Accountant',
    7: 'Scientist',
    8: 'Lawyer',
    9: 'Salesperson',
    10: 'Manager'
}

bmi_map = {
    0: 'Overweight',
    1: 'Normal',
    2: 'Obese',
    3: 'Normal Weight'
}

sleep_map = {
    0: 'Healthy',
    1: 'Sleep Apnea',
    2: 'Insomnia'
}
```

```python
new_dataset['Gender'] = new_dataset['Gender'].map(gender_map)

new_dataset['Occupation'] = new_dataset['Occupation'].map(occupation_map)

new_dataset['BMI Category'] = new_dataset['BMI Category'].map(bmi_map)

new_dataset['Sleep Disorder'] = new_dataset['Sleep Disorder'].map(sleep_map)
```
